```python
# ==========================================
# SALES DATA ANALYSIS USING NUMPY
# ==========================================

# Step 1: Import NumPy Library
import numpy as np

print("Supermarket Sales Data Analysis\n")

# ------------------------------------------
# Step 2: Create Dataset
# ------------------------------------------

# Product categories
products = np.array(["Rice", "Oil", "Milk", "Soap", "Sugar"])

# Quantity sold
quantity = np.array([100, 80, 150, 200, 120])

# Unit price (in Rupees)
unit_price = np.array([50, 120, 40, 30, 45])

# Profit per unit
profit_per_unit = np.array([5, 15, 4, 3, 6])

print("Products:", products)
print("Quantity Sold:", quantity)
print("Unit Price:", unit_price)
print("Profit per Unit:", profit_per_unit)

# ------------------------------------------
# Step 3: Broadcasting – Calculate Total Sales
# ------------------------------------------

total_sales = quantity * unit_price   # Broadcasting
print("\nTotal Sales for each product:", total_sales)

# ------------------------------------------
# Step 4: Calculate Total Revenue and Average Sales
# ------------------------------------------

total_revenue = np.sum(total_sales)
average_sales = np.mean(total_sales)

print("\nTotal Revenue:", total_revenue)
print("Average Sales:", average_sales)

# ------------------------------------------
# Step 5: Statistical Analysis
# ------------------------------------------

mean_sales = np.mean(total_sales)
median_sales = np.median(total_sales)
variance_sales = np.var(total_sales)
std_sales = np.std(total_sales)

print("\nStatistical Analysis:")
print("Mean:", mean_sales)
print("Median:", median_sales)
print("Variance:", variance_sales)
print("Standard Deviation:", std_sales)

# ------------------------------------------
# Step 6: Vectorized Computation – Profit Calculation
# ------------------------------------------

total_profit = quantity * profit_per_unit
overall_profit = np.sum(total_profit)

print("\nTotal Profit per Product:", total_profit)
print("Overall Profit:", overall_profit)

# ------------------------------------------
# Step 7: Indexing and Slicing
# ------------------------------------------

# First 3 products sales
print("\nFirst 3 Product Sales:", total_sales[:3])

# Sales of Milk
milk_index = np.where(products == "Milk")
print("Milk Sales:", total_sales[milk_index])

# Highest Sales Product
max_index = np.argmax(total_sales)
print("\nHighest Sales Product:", products[max_index])
print("Highest Sales Value:", total_sales[max_index])

print("\nAnalysis Completed Successfully!")
```
